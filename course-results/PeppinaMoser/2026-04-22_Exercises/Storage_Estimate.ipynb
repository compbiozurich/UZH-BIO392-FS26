# Storage Estimate for 1000 Genomes

## Assumptions
- 1000 samples
- Data types: FASTQ, BAM, VCF
- SAM not included (too large, not used for storage)
- FASTA not included (reference genome, stored once ~3 GB)

## WGS (per genome)
- FASTQ ≈ 100 GB
- BAM ≈ 100 GB
- VCF ≈ 1 GB

## WGS (1000 genomes)
- FASTQ: 100 GB × 1000 = 100 TB
- BAM: 100 GB × 1000 = 100 TB
- VCF: 1 GB × 1000 = 1 TB

**Total WGS ≈ 201 TB (~200 TB)**

## WES (per genome)
- FASTQ ≈ 10 GB
- BAM ≈ 5–10 GB
- VCF ≈ 0.1–0.5 GB

## WES (1000 genomes)
- FASTQ: 10 GB × 1000 = 10 TB
- BAM: 5–10 GB × 1000 = 5–10 TB
- VCF: 0.1–0.5 GB × 1000 = 0.1–0.5 TB

**Total WES ≈ 15–25 TB**
