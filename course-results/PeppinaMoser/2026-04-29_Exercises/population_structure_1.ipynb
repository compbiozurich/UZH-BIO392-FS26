## Note: ADMIXTURE Analysis

ADMIXTURE could not be executed in this environment due to

The expected output would be a stacked bar plot showing the 
ancestry proportions of each individual across K=8 inferred 
ancestral populations, grouped by population label 
(similar to the plot shown in the lecture slides).

The code for the analysis is provided below and would 
produce the ADMIXTURE plot if the tool were available.

```python
# ============================================================
# ADMIXTURE Analyse – Chromosom 1
# Füge diese Zellen nach deiner PCA ins Notebook ein
# ============================================================


# ── Schritt 1: ADMIXTURE installieren (einmalig im Terminal) ──
# conda install -c bioconda admixture


# ── Schritt 2: ADMIXTURE ausführen (K=8) ──
# K=8 bedeutet: 8 Abstammungsgruppen annehmen

# !admixture chr1_final.bed 8


# ── Schritt 3: Populationsinformation laden ──

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ADMIXTURE Output laden (.Q Datei = Abstammungsanteile pro Person)
Q = pd.read_csv("chr1_final.8.Q", sep=r"\s+", header=None)

# Populationsinformation vom 1000 Genomes Projekt laden
panel = pd.read_csv(
    "https://ftp.1000genomes.ebi.ac.uk/vol1/ftp/technical/working/20130606_sample_info/20130606_g1k.ped",
    sep="\t"
)

# Sample-IDs aus PLINK Datei laden
fam = pd.read_csv("chr1_final.fam", sep=r"\s+", header=None)
fam.columns = ["FID", "IID", "PAT", "MAT", "SEX", "PHENO"]


# ── Schritt 4: Daten zusammenführen & sortieren ──

# Populationsinfo mit ADMIXTURE Ergebnissen verbinden
Q["IID"] = fam["IID"].values
merged = Q.merge(
    panel[["Individual ID", "Population", "Super Population"]],
    left_on="IID",
    right_on="Individual ID"
)

# Nach Population sortieren (für schöne Gruppierung im Plot)
merged = merged.sort_values(["Super Population", "Population"])


# ── Schritt 5: ADMIXTURE Plot erstellen ──

fig, ax = plt.subplots(figsize=(14, 4))

colors = plt.cm.Set2.colors[:8]
bottom = np.zeros(len(merged))
q_cols = [c for c in merged.columns if isinstance(c, int)]

# Gestapelte Balken zeichnen
for i, col in enumerate(q_cols):
    ax.bar(range(len(merged)), merged[col], bottom=bottom,
           color=colors[i], width=1, linewidth=0)
    bottom += merged[col].values

# Populationsbeschriftung unten hinzufügen
pop_labels = merged["Population"].values
changes = [0] + [i + 1 for i in range(len(pop_labels) - 1)
                 if pop_labels[i] != pop_labels[i + 1]] + [len(pop_labels)]
for j in range(len(changes) - 1):
    mid = (changes[j] + changes[j + 1]) / 2
    ax.text(mid, -0.05, pop_labels[changes[j]],
            ha="center", va="top", fontsize=7, rotation=45)

ax.set_xlim(0, len(merged))
ax.set_ylim(0, 1)
ax.set_ylabel("Ancestry proportion")
ax.set_title("ADMIXTURE chr1 – K=8")
ax.axis("off")
plt.tight_layout()
plt.savefig("admixture_chr1_K8.png", dpi=150, bbox_inches="tight")
plt.show()
```
